### self
self is the first argument of every method in Python.
In Python, self is used as the first parameter in instance methods to refer to the current object.

```python
class Person:
    def say_hello():
        print(f"Hello! {self}")

p= Person() # => object
p.say_hello()
```

```python
class Person:
    def say_hello(self):
        print("Hello!")

p= Person() # => object
p.say_hello()
```

```python
#class
class Car:
    def __init__(self, color, brand): #initialization function (dunder)
        self.color = color # attribute
        self.brand = brand #attribute

    def start_engine(self): #method
        print(f"The {self.brand} car's engine started" )

    def describe(self): #method
        print(f"This car is a {self.color} {self.brand}")


#object
my_car = Car('red', 'Toyota')

#accessing attributes
print(my_car.color, my_car.brand)

#calling methods
my_car.start_engine()
my_car.describe()
```

## INHERITANCE

```python
class Employee:
    def __init__(self, name, Employee_id, salary):
        self.name = name
        self.Employee_id = Employee_id
        self.salary = salary

    def work(self):
        return (f'{self.name} is working')

    def get_details(self):
        print(f'name: {self.name}, Employee_id: {self.Employee_id}, salary: {self.salary}')
```

```python
employee = Employee('Amjad', 101, 50000)
```

```python
print(employee.name)
print(employee.salary)
print(employee.Employee_id)
print(employee.work())
employee.work()
print(employee.get_details()) 
employee.get_details()
#none print return wala case in methods when written in print and when not
```

```python
class Employee:
    def __init__(self, name, Employee_id, salary):
        self.name = name
        self.Employee_id = Employee_id
        self.salary = salary

    def work(self):
        print (f'{self.name} is working')

    def get_details(self):
        print(f'name: {self.name}, Employee_id: {self.Employee_id}, salary: {self.salary}')


#inheritance
class Developer(Employee):
    def __init__(self, name, Employee_id, salary, prog_lang):
        super().__init__(name, Employee_id, salary) # inheriting from parent
        self.prog_lang = prog_lang

    def code(self):
        print(f"{self.name} is coding in {self.prog_lang}")
```

```python
dev = Developer('Ali', 102, 60000, 'Python')
```

```python
print(dev.name)
print(dev.salary)
print(dev.Employee_id)
dev.work()
dev.get_details()
print(dev.prog_lang)
dev.code()
```

## POLYMORPHISM

Polymorphism means same function name (but different sigtnatures) being uses for different types.

```python
l = [1,2,3,4]
len(l)
```

```python
len('hello')
```

```python
type('123')
```

```python
type(123)
```

```python
class Cat:
    def sound(self):
        return "Meow"

class Dog:
    def sound(self):
        return "Woof"

# Polymorphism in action
for animal in (Cat(), Dog()):
    print(animal.sound())
```

## OVERLOADING

```python
class Greet():
    def welcome(self, name=''):
        print('welcome' + " "+ name)
```

```python
greet = Greet()
```

```python
greet.welcome()
```

```python
greet.welcome('SAM')
```

```python
class ATM:
    def withdraw(self, amount, currency='USD'):
        if currency == 'USD':
            print(f'withdrawing {amount} dollars.')
        elif currency == 'PKR':
            print(f'withdrawing {amount} rupees.')
        else:
            print('Currency not supported.')
        
```

```python
atm = ATM()
atm.withdraw(100)
atm.withdraw(100, 'PKR')
atm.withdraw(100, "USD")
atm.withdraw(100, 'INR')
```
